```python
import pandas as pd
import plotly.graph_objects as go
import json

# ==========================================
# 1. مرحلة قراءة وتنظيف الداتا الأساسية
# ==========================================
def load_and_clean_data(file_path):
    df = pd.read_excel(file_path)    # تحويل الكود لنص
    if 'الكود' in df.columns:
        df['الكود'] = df['الكود'].astype(str).str.replace('.0', '', regex=False)
    
    # ملء الفراغات بصفر (عشان نحافظ على شكل الداتا العريض)
    df = df.fillna(0)
    
    # إزالة أي مسافات من أسماء الأصناف
    if 'اسم الصنف ' in df.columns:
        df['اسم الصنف '] = df['اسم الصنف '].str.strip()
        
    return df

# ==========================================
# 2. دالة التحليل والرسم (اللي هتسلمها للباك إند)
# ==========================================
def plot_branch_performance(df):
    # هنستخرج أسماء الفروع من أسماء الأعمدة اللي بتنتهي بـ " مع"
    sales_cols = [col for col in df.columns if col.endswith(' مع')]
    stock_cols = [col for col in df.columns if col.endswith(' اس')]
    
    # نجمع إجمالي المبيعات والاستوك لكل فرع
    total_sales = df[sales_cols].sum().values
    total_stock = df[stock_cols].sum().values
    
    # تنظيف أسماء الفروع للرسم (نشيل كلمة " مع")
    branch_names = [col.replace(' مع', '') for col in sales_cols]
    
    # --- بداية استخدام Plotly للرسم ---
    fig = go.Figure()
    
    # عمود المبيعات
    fig.add_trace(go.Bar(
        x=branch_names, 
        y=total_sales, 
        name='المبيعات',
        marker_color='#2ecc71' # لون أخضر
    ))
    
    # عمود الاستوك
    fig.add_trace(go.Bar(
        x=branch_names, 
        y=total_stock, 
        name='الاستوك (المخزون)',
        marker_color='#3498db' # لون أزرق
    ))
    
    # تظبيط شكل الرسمة
    fig.update_layout(
        title='مقارنة المبيعات والمخزون لكل فرع',
        xaxis_title='اسم الفرع',
        yaxis_title='الكمية',
        barmode='group', # عشان العواميد تطلع جنب بعض مش فوق بعض
        template='plotly_white'
    )
    
    return fig

# ==========================================
# 3. مرحلة التجربة عندك (Local Testing)
# ==========================================
if __name__ == "__main__":
    # 1. قراءة الداتا (حط مسار الفايل بتاعك هنا)
    df_clean = load_and_clean_data('مبيعات + استوكات خلال فترة.xls')
    
    # 2. إنشاء الرسمة
    my_figure = plot_branch_performance(df_clean)
    
    # 3. عرض الرسمة عندك في المتصفح للتيست (Interactive Plot)
    my_figure.show() 
    
    # ==========================================
    # 4. إزاي الباك إند هيستخدمها (للعلم ليك)
    # ==========================================
    # الباك إند هيعمل كده عشان يبعتها للويب:
    # plot_json = my_figure.to_json()
    # return Response(plot_json, mimetype='application/json')
```
